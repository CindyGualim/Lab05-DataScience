### **Ejercicio 3:** Limpieza y Preprocesamiento de Datos

1. Convertir a minúsculas
2. Remover URLs
3. Remover menciones (@) y hashtags (#)
4. Remover emoticones
5. Remover puntuación
6. Remover números
7. Remover espacios extra
8. Tokenización
9. Remover stopwords (palabras comunes sin valor semántico)
10. Lemmatization (reducción a forma base)


```python
def limpiar_tweet(texto):


    # 1. Convertir a minúsculas
    texto = texto.lower()

    # 2. Remover URLs
    texto = re.sub(r'http\S+|www\S+|https\S+', '', texto, flags=re.MULTILINE)

    # 3. Remover menciones (@usuario) y hashtags (#tema)
    texto = re.sub(r'@\w+|#\w+', '', texto)

    # 4. Remover emoticones comunes
    emoticones = [':)', ':(', ':P', ':D', ':/', '://', ':*', ';)', ';(']
    for emoticono in emoticones:
        texto = texto.replace(emoticono, '')

    # 5. Remover puntuación
    texto = re.sub(f'[{re.escape(string.punctuation)}]', '', texto)

    # 6. Remover números
    # Nota: Se pueden mantener ciertos números como '911' si es relevante
    texto = re.sub(r'\d+', '', texto)

    # 7. Remover espacios extra
    texto = ' '.join(texto.split())

    # 8. Tokenización
    tokens = word_tokenize(texto)

    # 9. Remover stopwords (palabras comunes: the, is, at, etc.)
    stop_words = set(stopwords.words('english'))
    tokens = [token for token in tokens if token not in stop_words]

    # 10. Lemmatization (reducir palabras a su forma base)
    lemmatizer = WordNetLemmatizer()
    tokens = [lemmatizer.lemmatize(token) for token in tokens]

    return ' '.join(tokens)

print("✓ Función de limpieza definida")
```

```python
# Mostrar ejemplo de preprocesamiento
print("EJEMPLO DE PREPROCESAMIENTO:")
print("="*80)

# Seleccionar un tweet de ejemplo
idx_ejemplo = 0
texto_original = df['text'].iloc[idx_ejemplo]
texto_limpio = limpiar_tweet(texto_original)
target_original = df['target'].iloc[idx_ejemplo]

print(f"Tweet Original:\n{texto_original}")
print(f"\n" + "-"*80)
print(f"Tweet Limpio:\n{texto_limpio}")
print(f"\nTarget: {target_original} → {'DESASTRE REAL' if target_original == 1 else 'NO ES DESASTRE'}")
```

```python
# Aplicar preprocesamiento a todo el dataset
print("\nAplicando preprocesamiento a todos los tweets...")
print("Esto puede tomar 1-2 minutos...")

df['text_cleaned'] = df['text'].apply(limpiar_tweet)

print(f"✓ Preprocesamiento completado")
print(f"  {len(df):,} tweets procesados exitosamente")
print(f"  {df['text_cleaned'].isna().sum()} valores nulos")
```
