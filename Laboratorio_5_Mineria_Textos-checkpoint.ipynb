### **Ejercicio 3:** Limpieza y Preprocesamiento de Datos

1. Convertir a minúsculas
2. Remover URLs
3. Remover menciones (@) y hashtags (#)
4. Remover emoticones
5. Remover puntuación
6. Remover números
7. Remover espacios extra
8. Tokenización
9. Remover stopwords (palabras comunes sin valor semántico)
10. Lemmatization (reducción a forma base)


```python
def limpiar_tweet1(texto):


    # 1. Convertir a minúsculas
    texto = texto.lower()

    # 2. Remover URLs
    texto = re.sub(r'http\S+|www\S+|https\S+', '', texto, flags=re.MULTILINE)

    # 3. Remover menciones (@usuario) y hashtags (#tema)
    texto = re.sub(r'@\w+|#\w+', '', texto)

    # 4. Remover emoticones comunes
    emoticones = [':)', ':(', ':P', ':D', ':/', '://', ':*', ';)', ';(']
    for emoticono in emoticones:
        texto = texto.replace(emoticono, '')

    # 5. Remover puntuación
    texto = re.sub(f'[{re.escape(string.punctuation)}]', '', texto)

    # 6. Remover números
    # Nota: Se pueden mantener ciertos números como '911' si es relevante
    texto = re.sub(r'\d+', '', texto)

    # 7. Remover espacios extra
    texto = ' '.join(texto.split())

    # 8. Tokenización
    tokens = word_tokenize(texto)

    # 9. Remover stopwords (palabras comunes: the, is, at, etc.)
    stop_words = set(stopwords.words('english'))
    tokens = [token for token in tokens if token not in stop_words]

    # 10. Lemmatization (reducir palabras a su forma base)
    lemmatizer = WordNetLemmatizer()
    tokens = [lemmatizer.lemmatize(token) for token in tokens]

    return ' '.join(tokens)

print("Función de limpieza definida")
```

```python
def limpiar_tweet(texto):


    # 1. Convertir a minúsculas
    texto = texto.lower()

    # 2. Remover URLs
    texto = re.sub(r'http\S+|www\S+|https\S+', '', texto, flags=re.MULTILINE)

    # 3. Remover menciones (@usuario) 
    texto = re.sub(r'@\w++', '', texto)

    # 4. Remover emoticones comunes
    emoticones = [':)', ':(', ':P', ':D', ':/', '://', ':*', ';)', ';(']
    for emoticono in emoticones:
        texto = texto.replace(emoticono, '')

    # 5. Remover puntuación
    texto = re.sub(f'[{re.escape(string.punctuation)}]', '', texto)

    # 6. Remover números
    # Nota: Se pueden mantener ciertos números como '911' si es relevante
    texto = re.sub(r'\b(?!911\b)\d+', '', texto)

    # 7. Remover espacios extra
    texto = ' '.join(texto.split())

    # 8. Tokenización
    tokens = word_tokenize(texto)

    # 9. Remover stopwords (palabras comunes: the, is, at, etc.)
    stop_words = set(stopwords.words('english'))
    tokens = [token for token in tokens if token not in stop_words]

    # 10. Lemmatization (reducir palabras a su forma base)
    lemmatizer = WordNetLemmatizer()
    tokens = [lemmatizer.lemmatize(token) for token in tokens]

    return ' '.join(tokens)

print("Función de limpieza definida")
```

### 4. Frecuencia de palabras de tweets de desastre y tweets misceláneos

De primero se segmentarán los tweets en dos dataframe: uno de desastres y el otro de no desastres (misceláneos).
Luego, se extraerán las palabras, se tokenizaran y se contarán para ambos segmentos.
Finalmente se mostrarán las primeras 20 palabras más relevantes

```python
# 1. Segmentar el DataFrame por tipo de tweet
df_desastres = df[df['target'] == 1]
df_no_desastres = df[df['target'] == 0]

# 2. Extraer palabras, separarlas y contarlas (Desastres)
frecuencia_desastres = (df_desastres['text_cleaned']
                        .str.split()
                        .explode()
                        .value_counts()
                        .reset_index())
frecuencia_desastres.columns = ['Palabra', 'Frecuencia']

# 3. Extraer palabras, separarlas y contarlas (No Desastres)
frecuencia_no_desastres = (df_no_desastres['text_cleaned']
                           .str.split()
                           .explode()
                           .value_counts()
                           .reset_index())
frecuencia_no_desastres.columns = ['Palabra', 'Frecuencia']

# 4. Mostrar los resultados en formato de tabla (Top 15 de cada uno)
print("TOP 15 PALABRAS EN TWEETS DE DESASTRES REALES (Target = 1):")
display(frecuencia_desastres.head(20))

print("\n" + "="*50 + "\n")

print("TOP 15 PALABRAS EN TWEETS DE NO DESASTRES (Target = 0):")
display(frecuencia_no_desastres.head(20))
```

La palabra más repetida en cada categoría por sí sola aporta poco contexto: suelen ser palabras genéricas (por ejemplo *like*, *im*, *amp*) que sobreviven al filtrado de stopwords porque no están incluidas en la lista estándar de NLTK, pero que no distinguen si un tweet habla de un desastre real. Esto refuerza la necesidad de mirar más allá del top 1 y comparar las listas completas de frecuencias entre ambas categorías, como se hace a continuación.

```python
# Nube de palabras para cada categoría
texto_desastres = ' '.join(df_desastres['text_cleaned'])
texto_no_desastres = ' '.join(df_no_desastres['text_cleaned'])

wc_desastres = WordCloud(width=800, height=500, background_color='white',
                          colormap='Reds', max_words=100).generate(texto_desastres)
wc_no_desastres = WordCloud(width=800, height=500, background_color='white',
                             colormap='Blues', max_words=100).generate(texto_no_desastres)

fig, axes = plt.subplots(1, 2, figsize=(18, 8))

axes[0].imshow(wc_desastres, interpolation='bilinear')
axes[0].set_title('Nube de palabras: Tweets de Desastres Reales (target=1)', fontsize=13)
axes[0].axis('off')

axes[1].imshow(wc_no_desastres, interpolation='bilinear')
axes[1].set_title('Nube de palabras: Tweets No Desastres (target=0)', fontsize=13)
axes[1].axis('off')

plt.tight_layout()
plt.show()
```

```python
# Histograma con las palabras que más se repiten en cada categoría
top_n = 15

fig, axes = plt.subplots(1, 2, figsize=(16, 7))

sns.barplot(data=frecuencia_desastres.head(top_n), x='Frecuencia', y='Palabra',
            ax=axes[0], color='firebrick')
axes[0].set_title(f'Top {top_n} palabras - Tweets de Desastres Reales')
axes[0].set_xlabel('Frecuencia')
axes[0].set_ylabel('Palabra')

sns.barplot(data=frecuencia_no_desastres.head(top_n), x='Frecuencia', y='Palabra',
            ax=axes[1], color='steelblue')
axes[1].set_title(f'Top {top_n} palabras - Tweets No Desastres')
axes[1].set_xlabel('Frecuencia')
axes[1].set_ylabel('Palabra')

plt.tight_layout()
plt.show()
```

```python
# Palabras con presencia en ambas categorias
set_desastres = set(frecuencia_desastres['Palabra'])
set_no_desastres = set(frecuencia_no_desastres['Palabra'])
palabras_comunes = set_desastres & set_no_desastres

print(f"Palabras unicas en tweets de desastres: {len(set_desastres)}")
print(f"Palabras unicas en tweets no desastres: {len(set_no_desastres)}")
print(f"Palabras presentes en AMBAS categorias: {len(palabras_comunes)}")
print()

# Comparar la frecuencia de las palabras comunes mas repetidas en cada categoria
comparacion = (frecuencia_desastres[frecuencia_desastres['Palabra'].isin(palabras_comunes)]
               .merge(frecuencia_no_desastres[frecuencia_no_desastres['Palabra'].isin(palabras_comunes)],
                      on='Palabra', suffixes=('_desastre', '_no_desastre')))
comparacion['diferencia'] = comparacion['Frecuencia_desastre'] - comparacion['Frecuencia_no_desastre']
comparacion = comparacion.sort_values('Frecuencia_desastre', ascending=False)

print("TOP 15 PALABRAS COMUNES, ORDENADAS POR FRECUENCIA EN TWEETS DE DESASTRES:")
display(comparacion.head(15))
```

Un número considerable de palabras aparece en ambas categorías (por ejemplo *fire*, *new*, *people*, *via*), lo cual es esperable dado que el vocabulario de Twitter se comparte entre temas. Sin embargo, lo relevante no es su sola presencia sino la diferencia de frecuencia entre categorías: palabras como *fire* o *disaster* aparecen mucho más en tweets de desastres reales, mientras que términos genéricos como *like* o *im* están más balanceados entre ambas clases. Esto sugiere que, para el modelo de clasificación, conviene ponderar las palabras por su frecuencia relativa (p. ej. TF-IDF) en lugar de su sola aparición, ya que las palabras exclusivas o muy desbalanceadas entre clases serán las que aporten mayor poder discriminativo.
